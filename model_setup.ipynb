```python
from datetime import datetime
from collections import deque
import cv2
from ultralytics import YOLO
from ppe_nlp import generate_alert
from event_log import init_db, log_event   # ← new import

# ── Config ────────────────────────────────────────────────────────────────────
ZONE = "Zone A"
FRAME_BUFFER_SIZE = 10

# ── Load models ───────────────────────────────────────────────────────────────
pose_model = YOLO('/Users/mohitjain965405gmail.com/Downloads/ppe/yolo26s-pose.pt')
ppe_model  = YOLO('/Users/mohitjain965405gmail.com/Downloads/ppe/ppe_yolo.pt')

cap = cv2.VideoCapture('test1.mp4')

init_db()   # ← create table at startup if not exists

# ── Multi-frame validator state ───────────────────────────────────────────────
frame_buffers    = {}
confirmed_states = {}

def update_validator(person_id, current_violations: set) -> bool:
    if person_id not in frame_buffers:
        frame_buffers[person_id]    = deque(maxlen=FRAME_BUFFER_SIZE)
        confirmed_states[person_id] = None

    buffer = frame_buffers[person_id]
    buffer.append(frozenset(current_violations))

    if len(buffer) < FRAME_BUFFER_SIZE:
        return False

    if len(set(buffer)) != 1:
        return False

    consistent_state = current_violations

    if consistent_state != confirmed_states[person_id]:
        confirmed_states[person_id] = consistent_state
        return True

    return False


def check_near(point, objects):
    px, py = int(point[0]), int(point[1])
    for (x1, y1, x2, y2, conf) in objects:
        if x1 < px < x2 and y1 < py < y2:
            return True, conf
    return False, 0.0


# ── Main loop ─────────────────────────────────────────────────────────────────
while True:
    ret, frame = cap.read()
    if not ret:
        break

    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    pose_results = pose_model(frame, conf=0.4)
    ppe_results  = ppe_model(frame, conf=0.4)

    annotated = pose_results[0].plot()

    helmets, vests, gloves, boots = [], [], [], []

    for box in ppe_results[0].boxes:
        cls   = int(box.cls[0])
        label = ppe_model.names[cls]
        conf  = float(box.conf[0])
        x1, y1, x2, y2 = map(int, box.xyxy[0])

        if label == "helmet":
            helmets.append((x1, y1, x2, y2, conf))
        elif label == "vest":
            vests.append((x1, y1, x2, y2, conf))
        elif label == "gloves":
            gloves.append((x1, y1, x2, y2, conf))
        elif label == "boots":
            boots.append((x1, y1, x2, y2, conf))

    for r in pose_results:
        if r.keypoints is None:
            continue

        for kp_idx, person_kp in enumerate(r.keypoints.xy):
            kp = person_kp.tolist()

            person_id  = f"person_{kp_idx}"

            head       = kp[0]
            left_hand  = kp[9]
            right_hand = kp[10]
            left_foot  = kp[15]
            right_foot = kp[16]

            helmet_ok, helmet_conf = check_near(head, helmets)

            glove_left_ok,  glove_left_conf  = check_near(left_hand,  gloves)
            glove_right_ok, glove_right_conf = check_near(right_hand, gloves)
            glove_ok   = glove_left_ok or glove_right_ok
            glove_conf = max(glove_left_conf, glove_right_conf)

            boot_left_ok,  boot_left_conf  = check_near(left_foot,  boots)
            boot_right_ok, boot_right_conf = check_near(right_foot, boots)
            boot_ok   = boot_left_ok or boot_right_ok
            boot_conf = max(boot_left_conf, boot_right_conf)

            vest_ok, vest_conf = False, 0.0
            for (x1, y1, x2, y2, conf) in vests:
                cy = int((y1 + y2) / 2)
                if 100 < cy < frame.shape[0]:
                    vest_ok   = True
                    vest_conf = max(vest_conf, conf)

            violation_types = []
            violation_confs = []

            if not helmet_ok:
                violation_types.append("no_helmet")
            else:
                violation_confs.append(helmet_conf)

            if not vest_ok:
                violation_types.append("no_vest")
            else:
                violation_confs.append(vest_conf)

            if not glove_ok:
                violation_types.append("no_gloves")
            else:
                violation_confs.append(glove_conf)

            if not boot_ok:
                violation_types.append("no_boots")
            else:
                violation_confs.append(boot_conf)

            avg_conf = round(sum(violation_confs) / len(violation_confs), 2) if violation_confs else 0.0

            should_alert = update_validator(person_id, set(violation_types))

            if should_alert:
                event = {
                    "zone":       ZONE,
                    "timestamp":  timestamp,
                    "type":       violation_types,
                    "confidence": avg_conf
                }
                alert = generate_alert(event)
                log_event(alert)   # ← replaces print(), saves to DB + prints

            status_map = [
                ("Helmet" if helmet_ok else "No Helmet", helmet_ok),
                ("Vest"   if vest_ok   else "No Vest",   vest_ok),
                ("Gloves" if glove_ok  else "No Gloves", glove_ok),
                ("Boots"  if boot_ok   else "No Boots",  boot_ok),
            ]

            y_offset = 30 + (kp_idx * 130)
            for text, ok in status_map:
                color = (0, 255, 0) if ok else (0, 0, 255)
                cv2.putText(annotated, text, (20, y_offset),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2)
                y_offset += 30

    cv2.imshow("PPE Safety System", annotated)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
from datetime import datetime

def generate_alert(event):
    zone = event.get("zone", "Zone A")
    timestamp = event.get("timestamp")
    types = event.get("type", [])

    mapping = {
        "no_helmet": "helmet",
        "no_vest": "safety vest",
        "no_gloves": "gloves",
        "no_boots": "safety boots"
    }

    violations = [mapping[t] for t in types if t in mapping]

    if not violations:
        return f"All safety equipment is properly worn in {zone} at {timestamp}"

    elif len(violations) == 1:
        return f"Worker in {zone} is not wearing a {violations[0]} at {timestamp}"

    else:
        items = ", ".join(violations[:-1]) + " and " + violations[-1]
        return f"Worker in {zone} is not wearing {items} at {timestamp}"
    


import cv2
from ultralytics import YOLO

# Load models
pose_model = YOLO('/Users/mohitjain965405gmail.com/Downloads/ppe/yolo26s-pose.pt')
ppe_model = YOLO('/Users/mohitjain965405gmail.com/Downloads/ppe/ppe_yolo.pt')

cap = cv2.VideoCapture('test1.mp4')

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run both models
    pose_results = pose_model(frame, conf=0.4)
    ppe_results = ppe_model(frame, conf=0.4)

    annotated = pose_results[0].plot() 

    # Collect PPE detections
    helmets, vests, gloves, boots = [], [], [], []

    for box in ppe_results[0].boxes:
        cls = int(box.cls[0])
        label = ppe_model.names[cls]
        x1, y1, x2, y2 = map(int, box.xyxy[0])

        if label == "helmet":
            helmets.append((x1,y1,x2,y2))
        elif label == "vest":
            vests.append((x1,y1,x2,y2))
        elif label == "gloves":
            gloves.append((x1,y1,x2,y2))
        elif label == "boots":
            boots.append((x1,y1,x2,y2))

    # Process each person from pose model
    for r in pose_results:
        if r.keypoints is None:
            continue

        for person_kp in r.keypoints.xy:
            # Convert to list of keypoints
            kp = person_kp.tolist()

            # Keypoints indexes (YOLO pose)
            head = kp[0]        # nose
            left_hand = kp[9]
            right_hand = kp[10]
            left_foot = kp[15]
            right_foot = kp[16]

            def check_near(point, objects):
                px, py = int(point[0]), int(point[1])
                for (x1,y1,x2,y2) in objects:
                    if x1 < px < x2 and y1 < py < y2:
                        return True
                return False

            helmet_ok = check_near(head, helmets)
            glove_ok = check_near(left_hand, gloves) or check_near(right_hand, gloves)
            boot_ok = check_near(left_foot, boots) or check_near(right_foot, boots)

            # Vest check (center body)
            vest_ok = False
            for (x1,y1,x2,y2) in vests:
                cx = int((x1+x2)/2)
                cy = int((y1+y2)/2)
                if 100 < cy < frame.shape[0]:  # rough torso area
                    vest_ok = True

            # Draw result
            status_text = []

            if helmet_ok:
                status_text.append("Helmet")
            else:
                status_text.append("No Helmet")

            if vest_ok:
                status_text.append("Vest")
            else:
                status_text.append("No Vest")

            if glove_ok:
                status_text.append("Gloves")
            else:
                status_text.append("No Gloves")

            if boot_ok:
                status_text.append("Boots")
            else:
                status_text.append("No Boots")

            # Display text
            y_offset = 30
            for text in status_text:
                cv2.putText(annotated, text, (20, y_offset),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7,
                            (0,255,0) if "✔" in text else (0,0,255), 2)
                y_offset += 30

            # Console output (final result)
            print({
                "helmet": helmet_ok,
                "vest": vest_ok,
                "gloves": glove_ok,
                "boots": boot_ok
            })

    cv2.imshow("PPE Safety System", annotated)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
